# Omówienie regresji logistycznej

przykład regresji logistycznej dla jednej zmiennej objaśniającej: wieku.

```python
import math
import pandas as pd
from platform import python_version
from sklearn.linear_model import LogisticRegression
from sklearn import __version__ as sklearn_ver
from statsmodels import __version__ as statsmodels_ver

print(f'python: {python_version()}')
print(f'pandas: {pd.__version__}')
print(f'sklearn: {sklearn_ver}')
print(f'statsmodels: {statsmodels_ver}')
df = pd.read_excel('data.xlsx')
df.head(3)
X = df[['age']]
y = df['target']

model = LogisticRegression()
model.fit(X, y)

# probabilithy of death
age_to_predict = 30
predicted_probability = model.predict_proba([[age_to_predict]])[:, 1]

theta0 = model.intercept_[0]
theta1 = model.coef_[0][0]
print(f'theta1={round(theta1,3)}; theta0 ={round(theta0,3)}')

# probability of death from math formula
manual_probability = 1/(1+math.exp(-(age_to_predict*theta1+theta0)))
print(f'p(modelem)   = {predicted_probability[0]:.8f}')
print(f'p(wzorem)    = {manual_probability:.8f}')
print(f'roznica      = {abs(predicted_probability[0] - manual_probability):.8f}')
```

```python
import numpy as np, math
from sklearn.metrics import roc_auc_score, brier_score_loss
p = model.predict_proba(X)[:,1]
print("AUC:", roc_auc_score(y, p))
print("Brier:", brier_score_loss(y, p))
print("OR (per year):", math.exp(theta1))
age50_p = 1/(1+math.exp(-(theta0+theta1*50)))
print("P(Y=1 | age=50):", age50_p)
```

- `theta1` pokazuje, jak zmieniają się log-ilorazy szans po wzroście wieku o 1 rok.
- `exp(theta1)` to iloraz szans dla zmiany wieku o 1 jednostkę.
- Zgodność wyniku `predict_proba()` ze wzorem sigmoidalnym potwierdza, że model realizuje klasyczną regresję logistyczną.
